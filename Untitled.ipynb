Converting transaction_date to date (datatype change)

```python
df['Transaction_Date'] =pd.to_datetime(df['Transaction_Date'])
```

# Feature Engineering

```python
df['Year'] = df['Transaction_Date'].dt.year
```

```python
df['Month'] =df['Transaction_Date'].dt.month
df['Day'] =df['Transaction_Date'].dt.day
df['Day_Name'] =df['Transaction_Date'].dt.day_name()
```

# EDA

```python
# Total Transaction
df['Transaction_ID'].count()
```

```python
# Total Revenue
df['Amount_NPR'].sum()
```

```python
# Transaction Status Analysis
status_analysis = df['Transaction_Status'].value_counts().reset_index()
status_analysis.columns = ['Status', 'Count']
status_analysis
```

```python
# Top Cities
df['City'].value_counts()
```

```python
# Payment Method Usage
df['Payment_Method'].value_counts()
```

```python
# Merchant Category Performance
user_analysis = df.groupby('Merchant_Category')['Amount_NPR'].sum().sort_values(ascending=False).reset_index()
user_analysis.columns = ['Merchant', 'Total_Amount']
user_analysis
```

```python
# Monthly Trend
df.groupby('Month')['Amount_NPR'].sum().sort_values(ascending=False)
```

```python
# Device Usage
df['Device_Type'].value_counts()
```

```python
# User Behavior
df.groupby('User_ID')['Amount_NPR'].sum().sort_values(ascending=False)
```

```python
# Peak transaction Day
df['Day_Name'].value_counts()
```

```python
# Failure Rate
failure_rate = df[df['Transaction_Status'] == 'Failed'].shape[0] / df.shape[0]
print(failure_rate)
```

```python
# Monthly Revenue
monthly_revenue = df.groupby(['Year', 'Month'])['Amount_NPR'].sum().reset_index()
monthly_revenue
```

```python
# City_wise Transaction
City_analysis = df.groupby('City').agg({'Transaction_ID' : 'count', 'Amount_NPR' : 'sum' }).reset_index()
City_analysis.rename(columns = {'Transaction_ID' : 'Total_Transaction'}, inplace = True)
City_analysis
```

```python
# Payment Method Analysis
payment_analysis = df.groupby('Payment_Method').agg({ 'Transaction_ID': 'count', 'Amount_NPR' : 'sum'}).reset_index()
payment_analysis.rename(columns = {'Transaction_ID' : 'Total_Transaction'}, inplace = True)
payment_analysis
```

```python
# Merchant Category Analysis
merchant_analysis = df.groupby('Merchant_Category')['Amount_NPR'].sum().reset_index()
merchant_analysis
```

```python
# High Value Transaction
high_value = df[df['Amount_NPR']> 5000]
high_value
```

# Visualisation

```python
# Transaction Status (Success vs Failed)
sns.countplot(x='Transaction_Status', data=df)
plt.title('Transation Status Distribution')
plt.show()
```

```python
# Amount Distribution
sns.histplot(df['Amount_NPR'], kde=True)
plt.title("Transaction Amount Distribution")
plt.show()
```

```python
# Monthly Trend
monthly = df.groupby('Month')['Amount_NPR'].sum().reset_index()

sns.lineplot(x='Month', y='Amount_NPR', data= monthly)
plt.title("Monthly Transaction Trend")
plt.show()
```

```python
# Top Cities
top_cities = df['City'].value_counts().head(10)

sns.barplot(x= top_cities.values , y=top_cities.index)
plt.title("Top Cities by Transactions")
plt.show()
```

```python
# Payment Method Usage
sns.countplot(x='Payment_Method', data=df)
plt.title("Payment Method Usage")
plt.xticks(rotation=20)
plt.show()
```

```python
# Merchant Category Revenue
merchant = df.groupby('Merchant_Category')['Amount_NPR'].sum().sort_values(ascending=False)
sns.barplot(x=merchant.values, y=merchant.index)
plt.title("Revenue by Merchant Category")
plt.show()
```

```python
sns.countplot(x='Device_Type', data=df)
plt.title("Device Usage")
plt.show()
```

```python
# Which method is used for high-value transactions
sns.boxplot(x='Payment_Method', y='Amount_NPR', data=df)
plt.title("Transaction Amount by Payment Method")
plt.show()
```

```python
# Day wise Activities
sns.countplot(x='Day_Name', data=df, order=[
    'Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'
])
plt.title("Transactions by Day")
plt.xticks(rotation=30)
plt.show()
```
